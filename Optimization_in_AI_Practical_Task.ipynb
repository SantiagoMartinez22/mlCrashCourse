# 🧠 **Task 1 — Resource Scheduling (Greedy)**

### 📌 Умова

* Є задачі (час виконання + які ресурси потрібні)
* Є ресурси з обмеженою місткістю
* Треба розподілити задачі так, щоб:

  * вистачило ресурсів
  * вкластися в дедлайн

---

### 💡 Суть алгоритму (Greedy)

Алгоритм діє локально:
👉 **на кожному кроці обирає найпростішу (найкоротшу) задачу і намагається швидко її “прилаштувати”**

Він **не шукає ідеальне рішення глобально**, а будує його крок за кроком:

* “що найвигідніше зробити прямо зараз?”

---

### 🧩 Як він працює

1. Спочатку впорядковує задачі:

   * короткі → першими (їх легше “втиснути” в дедлайн)

2. Далі дивиться на ресурси:

   * починає з тих, що простіше заповнити

3. Для кожної задачі:

   * перевіряє:

     * чи підходить ресурс
     * чи вистачає місткості
     * чи не вилітаємо за дедлайн
   * якщо так → одразу призначає

4. Після призначення:

   * ресурс “зменшується”
   * час накопичується

---

### ⚠️ Інтуїція

> “Спочатку швидко закриваємо маленькі задачі, поки є ресурси і час”

# 🚚 **Task 2 — Vehicle Routing Problem (Brute Force)**

### 📌 Умова

* Є склад (старт)
* Є клієнти (точки)
* Є машини з обмеженням
* Треба мінімізувати загальну відстань

---

### 💡 Суть алгоритму (Brute Force)

Алгоритм **не думає — він перебирає ВСЕ**

👉 Ідея:

> “Я перевірю кожен можливий варіант і просто візьму найкращий”

---

### 🧩 Як він працює

1. Генерує всі можливі порядки клієнтів
   (усі варіанти, як можна їх відвідати)

2. Кожен варіант:

   * ділить на частини (по кількості клієнтів на машину)
   * це маршрути різних машин

3. Для кожного маршруту:

   * рахує відстань між сусідніми точками

4. Складає загальну відстань

5. Порівнює всі варіанти:

   * обирає мінімальний

---

### ⚠️ Інтуїція

> “Я не шукаю розумно — я гарантовано знайду найкраще, бо перевірю все”

# 📦 **Task 3 — Inventory Management (Linear Programming)**

### 📌 Умова

* Є попит по періодах
* Є витрати:

  * зберігання
  * замовлення
* Треба:

  * не мати дефіциту
  * не тримати зайвий товар
  * мінімізувати витрати

---

### 💡 Суть алгоритму (Linear Programming)

Алгоритм не “перебирає” і не “жадібний”

👉 Він:

> будує **математичну модель задачі** і знаходить найкраще рішення через оптимізацію

---

### 🧩 Як він працює

1. Визначає змінні:

   * скільки товару є
   * скільки замовляємо

2. Формує ціль:

   * мінімізувати витрати (зберігання + замовлення)

3. Додає логіку у вигляді обмежень:

   * запас має покривати попит
   * враховується попередній період (баланс)
   * запас не перевищує допустимий рівень
   * значення не можуть бути від’ємними

4. Передає це “солверу”

   * він математично знаходить оптимум

---

### ⚠️ Інтуїція

> “Я описую задачу мовою математики, а система сама знаходить найкраще рішення”

TASK 1

```python
# ==============================
# Task 1 - Resource Scheduling (Greedy)
# ==============================

# Tasks: execution time + required resources
tasks = [
    {"id": "T1", "time": 2, "resource_need": 1},
    {"id": "T2", "time": 5, "resource_need": 2},
    {"id": "T3", "time": 1, "resource_need": 1},
    {"id": "T4", "time": 3, "resource_need": 1}
]

# Resources with limited capacity
resources = [
    {"id": "R1", "capacity": 3},
    {"id": "R2", "capacity": 2}
]

deadline = 8


def greedy_resource_scheduling(tasks, resources, deadline):

    # Sort tasks by shortest execution time first
    tasks = sorted(tasks, key=lambda task: task["time"])

    assignments = []

    # Track accumulated execution time per resource
    resource_time = {
        resource["id"]: 0
        for resource in resources
    }

    # Track remaining capacity per resource
    resource_capacity = {
        resource["id"]: resource["capacity"]
        for resource in resources
    }

    # Greedy assignment
    for task in tasks:

        assigned = False

        for resource in resources:

            resource_id = resource["id"]

            enough_capacity = (
                resource_capacity[resource_id]
                >= task["resource_need"]
            )

            within_deadline = (
                resource_time[resource_id]
                + task["time"]
                <= deadline
            )

            if enough_capacity and within_deadline:

                assignments.append(
                    (task["id"], resource_id)
                )

                resource_capacity[resource_id] -= task["resource_need"]
                resource_time[resource_id] += task["time"]

                assigned = True
                break

        if not assigned:
            assignments.append(
                (task["id"], "Not Assigned")
            )

    return assignments


# Run algorithm
result = greedy_resource_scheduling(
    tasks,
    resources,
    deadline
)

print("Greedy Resource Scheduling Results")
print("-" * 40)

for task, resource in result:
    print(f"{task} -> {resource}")

print("\nRemaining Capacities:")
for resource in resources:
    rid = resource["id"]
    print(f"{rid}: {resource['capacity']} initial capacity")
```

```python
# ==========================================
# Task 2 - Vehicle Routing Problem (Brute Force)
# ==========================================

from itertools import permutations
from math import sqrt

# Warehouse (starting point)
warehouse = (0, 0)

# Customers coordinates
customers = {
    "C1": (2, 3),
    "C2": (5, 4),
    "C3": (1, 6),
    "C4": (6, 1)
}

# Number of vehicles
num_vehicles = 2


# Euclidean distance
def distance(p1, p2):
    return sqrt((p1[0] - p2[0])**2 + (p1[1] - p2[1])**2)


# Route distance (warehouse -> customers -> warehouse)
def route_distance(route):

    total = 0
    current = warehouse

    for customer in route:
        total += distance(current, customers[customer])
        current = customers[customer]

    total += distance(current, warehouse)

    return total


best_distance = float("inf")
best_routes = None

customer_list = list(customers.keys())

# Generate ALL possible customer visit orders
for perm in permutations(customer_list):

    # Split customers between vehicles
    split_point = len(customer_list) // num_vehicles

    vehicle_1 = perm[:split_point]
    vehicle_2 = perm[split_point:]

    total_distance = (
        route_distance(vehicle_1)
        + route_distance(vehicle_2)
    )

    if total_distance < best_distance:
        best_distance = total_distance
        best_routes = (vehicle_1, vehicle_2)


# Results
print("Vehicle Routing Problem (Brute Force)")
print("-" * 50)

print("Best Route for Vehicle 1:")
print("Warehouse ->", " -> ".join(best_routes[0]), "-> Warehouse")

print("\nBest Route for Vehicle 2:")
print("Warehouse ->", " -> ".join(best_routes[1]), "-> Warehouse")

print("\nMinimum Total Distance:")
print(round(best_distance, 2))
```

```python
# ==========================================
# Task 3 - Inventory Management (Linear Programming)
# ==========================================

# Install PuLP if needed
!pip -q install pulp

from pulp import *

# Demand for each period
demand = [20, 30, 25, 35]

# Costs
holding_cost = 2      # storage cost per unit
ordering_cost = 5     # ordering cost per unit

# Maximum inventory allowed
max_inventory = 50

# Number of periods
n = len(demand)

# Create optimization model
model = LpProblem(
    "Inventory_Management",
    LpMinimize
)

# Decision variables
order = [
    LpVariable(f"Order_{t}", lowBound=0)
    for t in range(n)
]

inventory = [
    LpVariable(
        f"Inventory_{t}",
        lowBound=0,
        upBound=max_inventory
    )
    for t in range(n)
]

# Objective function:
# Minimize ordering cost + holding cost
model += lpSum(
    ordering_cost * order[t]
    + holding_cost * inventory[t]
    for t in range(n)
)

# Inventory balance constraints
for t in range(n):

    if t == 0:
        model += (
            order[t] - demand[t]
            == inventory[t]
        )
    else:
        model += (
            inventory[t - 1]
            + order[t]
            - demand[t]
            == inventory[t]
        )

# Solve the optimization problem
model.solve()

# Results
print("Inventory Management (Linear Programming)")
print("-" * 50)

for t in range(n):
    print(
        f"Period {t+1}: "
        f"Order = {order[t].varValue:.0f}, "
        f"Inventory = {inventory[t].varValue:.0f}"
    )

print("\nMinimum Total Cost:")
print(value(model.objective))
```
